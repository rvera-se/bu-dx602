````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def mean(values):
    '''Returns the arithmetic mean of a list of numbers.'''

    # Initialize total
    total = 0

    # Loop that iterates through the values list and computes the total
    for i in values:
        total += i

    # Calculate the mean
    avg = total / len(values)

    # Returns the mean of the list of values (total / length of values list)
    return total / len(values)
````

**Concept:** *Accumulator pattern with numeric result, list iteration.*

**Task:**

Write a function `variance(values)` that returns the population variance of a list of numbers using this formula:

$$
\text{variance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2
$$

Rules:

* Must use loops + accumulator (no `sum()`).

* No imports.

* Must call your `mean(values)` function to get the mean.

* Then run the provided test case and print in the same format.

**Sample Output:**

```

mean: 6.0
variance: 4.0
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def variance(values):
    '''Returns the population variance of a list of numbers using the specified variance formula'''

    # Initialize total
    total = 0

    # Calculate mean of values
    avg = mean(values)

    # Loop that iterates through each integer in values to calculate
    # the sum of the squares of the subtraction of the mean from each value
    # And adds it to the total
    for i in values:
        total += (i - avg)**2

    # Returns the variance of the list of values
    return total / len(values)
````

**Problem 8**

**Concept:** *Accumulator pattern with numeric result, list iteration, index-based loop.*

**Task:**

Write a function `covariance(xs, ys)` that computes the population covariance between two series:

$$
\text{covariance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})
$$

Rules:

* If `xs` and `ys` are not the same length, print:
    
    `xs and ys must be the same length`
    
    and return `0`.

* Must use an index-based loop.

* Must use your `mean()` function.

* Then run the given test case and print exactly in the same format.

**Sample Output:**

```
x values: [6, 3, 7, 9, 5]
y values: [4, 1, 7, 5, 0]
covariance: 3.8
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def covariance(xs, ys):
    '''Computes the population covariance between two series.'''

    # If the lists are not the same length, prints an error message and returns zero
    if(len(xs) != len(ys)):
        print('xs and ys must be the same length')
        return 0

    # Initialize total
    total = 0

    # Calculate the means
    xs_mean = mean(xs)
    ys_mean = mean(ys)

    # Iterates through each value in the xs and ys list and calculates the sum of
    # The multiplications of the deviation of each item in xs times the deviation of each item in ys
    for i in range(0, len(xs)):
        total += (xs[i] - xs_mean) * (ys[i] - ys_mean)

    # Returns the covariance
    return total / len(xs)
````

**Problem 12**

**Concept:** *Accumulator pattern to process a string, produce a string result.*

**Task:**

Write a function `cipher(s, n)` that applies a Caesar cipher shift to a plaintext string.

Rules:

* Assume lowercase letters only for shifting (`a` to `z`).

* Do not shift non-letters (spaces stay spaces, punctuation stays punctuation).

* Must wrap around the alphabet if shifting past `z` or before `a`.

* Must use an accumulator string.

* Then run the exact test cases shown and print in the same format.

**Sample Output:**

```
plaintext = abc, ciphered = def
plaintext = abc xyz, ciphered = def abc
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def cipher(s, n):
    '''Applies a Caesar cipher shift of distance n to a plaintext string s.'''

    # Initialize accumulator
    shifted_text = ""

    # Iterates through each character in string s to apply the Caesar shift of n
    for char in s:

        # If the ASCII code refers to a character from a to z, apply the shift
        if (ord(char) >= 97 and ord(char) <= 122):

            # If the shift will push the character past z, wrap around from a
            if ((ord(char) + n) > 122):
                shifted_text += chr(ord(char) + n - 26)

            # Else if the shift will push the character before a, wrap around from z
            elif ((ord(char) + n) < 97):
                shifted_text += chr(ord(char) + n + 26)

            # Otherwise, only apply the shift
            else:
                shifted_text += chr(ord(char) + n)

        # Leave character as is if it is not a lowercase a-z
        else:
            shifted_text += char

    # Returns the new ciphered string
    return shifted_text 

    
````
