**Review Problem 2**

**Concept:** *Processing a list with a definite loop, using accumulator pattern.*

**Task:**

Write a function `total_length(words)` that returns the sum of the lengths of all words in a list using the accumulator pattern.

Write a second version using list comprehension.

**Sample Output:**

```
total_length(['data', 'science', 'rocks'])
```
should return `15`




`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def total_length(words):
    '''Returns the sum of the lengths of all words in a list using the accumulator pattern.'''
    # Initialize accumulator
    sum = 0

    # Iterate through words in 'words' and add their length to the accumulator
    for w in words:
        sum += len(w)

    # Return sum
    return sum

def total_lengthV2(words):
    '''Does the same as total_length but using a list comprehension instead.'''

    # Return statement with list comprehension.
    return sum([len(w) for w in words])
`````

**Review Problem 3**

**Concept:** *List processing, accumulation, decision statements, creating a new list as result.*

**Task:**

Write a function `union(a, b)` that returns a new list containing all unique elements from both lists.

Rules:

* Cannot use a set

* Cannot modify input lists

* Must create a new list

* Order does not matter

**Sample Output:**

```
union of [1] with [1,2] should return [1, 2]
union of [1, 3] with [2, 4, 5] should return [1, 2, 3, 4, 5]
union of [4, 5] [1, 2, 3, 4] should return [4, 5, 1, 2, 3]
```




`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def union(a, b):
    '''Returns a new list containing all unique elements from both lists.'''
    # Initialize accumulator
    unifiedList = []

    # Iterates through elements 'e' in list 'a' and adds them to unifiedList
    for e in a:
        unifiedList.append(e)

    # Iterates through elements 'e' in list 'b' and adds them to unifiedList if they are not in unifiedList.
    for e in b:
        if(e not in unifiedList): unifiedList.append(e)

    # Returns new list
    return unifiedList
`````

**Problem 4**

**Concept:** *Using `set` to remove duplicates.*

**Task:**

Write a function `union_2(a, b)` that returns a list of unique elements by converting the combined list into a set and back into a list.

**Sample Output:**

```
union of [1] with [1,2] should return [1, 2]
union of [1, 3] with [2, 4, 5] should return [1, 2, 3, 4, 5]
union of [4, 5] [1, 2, 3, 4] should return [4, 5, 1, 2, 3]
```


`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def union_2(a, b):
    '''Returns a list of unique elements by converting the combined list into a set and back into a list.'''
    # Initialize sets and accumulator
    a_set = set(a)
    b_set = set(b)
    combinedSetList = []

    # Unites sets a_set and b_set using the OR '|' operator for sets
    combinedSet = a_set | b_set

    # Iterates through combinedSet and adds its elements into the new list
    for e in combinedSet:
        combinedSetList.append(e)

    # Returns the new list
    return combinedSetList
`````

`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def unique_letters(s):
    '''Returns a list of unique characters in a string using list comprehension and set'''
    # Create set of unique characters in string 's'
    unique_chars = set(s)

    # Add the characters in unique_chars to a new list using list comprehension and return it
    return [c for c in unique_chars]
`````

**Problem 7**

**Task:**

Write a function `word_score(word)` that returns the Scrabble score of a word using the dictionary `scrabble_scores`.

**Sample Output:**
```

```
should return
```
cat: 5
```




`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def word_score(word):
    '''Returns the Scrabble score of a word using the dictionary scrabble_scores.'''

    # Initialize accumulator
    score = 0

    # Iterate through the characters in the word
    for chr in word:

        # Calculate the score of each character and add to accumulator if the character is valid
        if (chr in scrabble_scores): score += scrabble_scores[chr]

    # Return score
    return score
`````

**Problem 9**

**Concept:** *Using list data type, dictionary data type, loop, dictionary element access, accumulator pattern.*

**Task:**

Write a function `total_points(games_results, team_name)` that returns the total points scored by a team across all games.

**Sample Output:**
```
games_results = [
    {'A': 5, 'B': 7},
    {'A': 10, 'C': 3}
]
print(total_points(games_results, 'A'))
```
should return
```
total points for A: 15
```



`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def total_points(games_results, team_name):
    '''Returns the total points scored by a team across all games.'''

    # Initialize accumulator
    points = 0

    # Iterate through the game results
    for game in games_results:

        # If the specified team_name is present in a game's results, add the points scored to accumulator
        if (team_name in game): points += game[team_name]

    # Return total points
    return points
`````

`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def sort_games_by_score(games, team_name):
    '''Returns a new list sorted by a team's score in descending order.'''

    # Initialize a new list with the values sorted
    sorted_games = sorted(games, key=lambda x: x[team_name], reverse = True)

    # Return new list
    return sorted_games
`````

**Problem 11**

**Concept:** *Using dictionary data type, loop, decision statements.*

**Task:**

Write a function `best_score(games, team_name)` that returns the game dictionary where the team scored the most points.

**Sample Output:**
```
games = [
    {'A': 8, 'B': 10},
    {'A': 15, 'C': 7},
    {'A': 12, 'D': 9}
]
print(best_score(games, 'A'))
```
should return
```
{'A': 15, 'C': 7}
```




`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def best_score(games, team_name):
    '''Returns the game dictionary where the team scored the most points.'''

    # Initialize new dictionary
    most_points = {team_name : 0}

    # Iterate through games
    for game in games:

        # If team_name scored higher points in current game, change the pointer of most_points to that game
        if (game[team_name] > most_points[team_name]):
            most_points = game

    # Return dictionary where team scored most points
    return most_points
`````

**Problem 12**

**Concept:** *Using list data type, dictionary data type, sorting.*

**Task:**

Write a function `best_word(words)` that returns the word with the highest Scrabble score.

**Sample Output:**
```
best_word(['dog', 'quiz', 'cat'])
```
should return `quiz`, because `quiz` has the highest scrabble score.




`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def best_word(words):
    '''Returns the word with the highest Scrabble score.'''

    # Create new dictionary using words from 'words' and their scores
    wordsAndScores = {word: word_score(word) for word in words}

    # Sorth the dictionary by word scores in descending order
    wordsAndScores = dict(sorted(wordsAndScores.items(), key=lambda x: x[1], reverse = True))

    # Return first word with highest score
    return next(iter(wordsAndScores))
`````
